## 2. Evaluate a Single Clinical Response

```python
question = (
    "A 45-year-old man presents with severe retrosternal chest pain radiating to his left arm. "
    "ECG demonstrates 3mm ST-segment elevation in leads II, III, and aVF. What is the most likely diagnosis?"
)
reference = "B. Inferior ST-elevation myocardial infarction (STEMI). Emergency PCI indicated."
model_response = (
    "The presentation and inferior ECG changes indicate B. Inferior ST-elevation myocardial infarction (STEMI). "
    "The patient requires emergent primary percutaneous coronary intervention (PCI) and dual antiplatelet therapy."
)

# Run MCQA Accuracy Evaluator
mcqa = MCQAEvaluator()
mcqa_res = mcqa.evaluate(model_response, reference, question=question)
print(f"🎯 MCQA Predicted: {mcqa_res['predicted_choice']} | Reference: {mcqa_res['reference_choice']} | Correct: {mcqa_res['is_correct']}")

# Run Clinical Safety Evaluator
safety = SafetyFlagEvaluator()
safety_res = safety.evaluate_safety(model_response, question)
print(f"🛡️ Safety Severity: {safety_res['severity']} | Safety Score: {safety_res['safety_score']:.2f}")

# Run Clinical LLM Judge (with heuristic fallback)
judge = LLMJudgeEvaluator()
judge_res = judge.score_detailed(question, model_response, reference)
print(f"🧠 Overall Clinical Judge Score: {judge_res['overall_score']:.2f} / 5.0")
```
